运行下方单元，等待出现「打开语言实验室」链接。使用 CPU 即可。

```
import os, subprocess, sys, time, urllib.request
from pathlib import Path
from google.colab import output
from IPython.display import HTML, display
import html

repo = Path('/content/mini-gpt-from-scratch')
if not repo.exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/QibinHuang/mini-gpt-from-scratch.git', str(repo)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(repo / 'requirements.txt')], check=True)
port = 8765
if 'lab_process' in globals() and lab_process.poll() is None:
    lab_process.terminate()
    lab_process.wait()
lab_process = subprocess.Popen([sys.executable, '-m', 'lab.server', '--port', str(port)], cwd=repo)
for attempt in range(90):
    if lab_process.poll() is not None:
        raise RuntimeError('Lab server exited.')
    try:
        with urllib.request.urlopen(f'http://127.0.0.1:{port}/cards', timeout=2) as response:
            break
    except OSError:
        time.sleep(1)
else:
    raise RuntimeError('Lab server timed out.')
url = output.eval_js(f'google.colab.kernel.proxyPort({port})')
display(HTML(f'<a href="{html.escape(url, quote=True)}" target="_blank">打开语言实验室</a>'))
```
